### Try - except

```python
# Ejercico que solicita 2 numeros y los suma.
# Necesitamos enviar un mensaje al usuario cuando introduzca una letra
# en la variable num1 como podemos hacer esto?
#con los cambios minimos

print("Vamos a sumar 2 numeros")

num1 = input("Introduce el primer numero: ")
num2 = input("Introduce el segundo numero: ")

try:
    num1 = int(num1)
except:
    num1 = input("No has elegido un numero, elige un numero: ")

num1 = int(num1)   

num2 = int(num2)

suma = num1 + num2

print("La suma de ambos numeros es: ", suma)
```

```python
try:
    5 + 5
except:
        print("No se ha podido realizar la operacion")
```

```python
# Hacer un programa que nos salude si tenemos elegimos 1:

print("Este programa tiene varias opciones elige del 1 al 10 y disfruta")
try:
    num = int(input("Elige el numero 1 si quieres que te saluden: "))
    if num == 1:
        print("Hola")
    elif num == 2:
        print("adios")
    else:
        print("Seguimos trabajando en ello")
except:
    print(" No has elegido un numero:")

```

### Bucle While

```python
repetir = True
num1 = input("Cual es el numero? ")
while repetir == True:
    try: 
        num1 = int(num1)
        #repetir = False
    except:
        num1 = int(input("No has elegido un numero, elige un numero: "))
print("El numero es:, ", num1)
    
```

```python
repetir = True

num1 = int(input("Cual es el numero? "))   

while repetir == True:
    if num1 == 1:
        repetir = False  
    elif num1 != 1:
        num1 = int(input(" Si quieres salir escribe 1" ))
        
print("El numero es: ", num1)
    
```

```python
print("Programa que deletrea ")
nombre = input("Escribe un nombre")

for letra in range(len(nombre)):
    print(nombre[letra])
```
