## 2. Prueba de escritorio: piensa como Python 🟢

**Conceptos:** `range`, `while`, `break`, `continue`, excepciones

**Sin ejecutar nada**, escribe en la tabla qué imprime cada fragmento. Luego ejecútalo en la celda de abajo y explica **cada** diferencia entre tu predicción y el resultado real.

| # | Código | Mi predicción | Resultado real | ¿Por qué? |
|---|---|---|---|---|
| a | `for i in range(3): print(i)` | | | |
| b | `for i in range(2, 10, 3): print(i)` | | | |
| c | `for i in range(5, 0, -2): print(i)` | | | |
| d | `for i in range(5, 0): print(i)` | | | |
| e | `x = 10` → `while x > 0: x = x - 3` → `print(x)` | | | |
| f | Recorrer `"Claro"` e imprimir cada letra, con `break` al llegar a la `"a"` | | | |
| g | Recorrer `"Claro"` y saltar las vocales con `continue` | | | |
| h | Sumar los pares del 1 al 10 | | | |
| i | `int("3.5")` dentro de un `try` | | | |
| j | `10 / 0` con dos `except` | | | |
| k | Triángulo con `for` anidados | | | |

**Pregunta de cierre:** el siguiente código **no está en la celda** porque nunca termina. Explica por qué y corrígelo cambiando **un solo carácter**:
```python
n = 5
while n != 0:
    n = n - 2
```

````python
# Verifica tus predicciones (solo DESPUÉS de llenar la tabla)
print("--- a")
for i in range(3):
    print(i)

print("--- b")
for i in range(2, 10, 3):
    print(i)

print("--- c")
for i in range(5, 0, -2):
    print(i)

print("--- d")
for i in range(5, 0):
    print(i)

print("--- e")
x = 10
while x > 0:
    x = x - 3
print(x)

print("--- f")
for letra in "Claro":
    if letra == "a":
        break
    print(letra)

print("--- g")
for letra in "Claro":
    if letra in "aeiou":
        continue
    print(letra, end="")
print()

print("--- h")
suma = 0
for i in range(1, 11):
    if i % 2 == 0:
        suma = suma + i
print(suma)

print("--- i")
try:
    print(int("3.5"))
except ValueError:
    print("Error de valor")

print("--- j")
try:
    print(10 / 0)
except ValueError:
    print("A")
except ZeroDivisionError:
    print("B")

print("--- k")
for i in range(4):
    for j in range(i):
        print("*", end="")
    print()
````
